```python
import pandas as pd
import numpy as np
import os

def standardize_variant_dataset(input_path, output_parquet):
    print(f"========================================================")
    print(f"🛠️ ĐANG CHUẨN HÓA TẬP DỮ LIỆU: {os.path.basename(input_path)}")
    print(f"========================================================")
    
    # 1. ĐỌC DỮ LIỆU TỰ ĐỘNG THEO ĐUÔI FILE
    if input_path.endswith('.csv'):
        df = pd.read_csv(input_path, low_memory=False)
    elif input_path.endswith('.tsv') or input_path.endswith('.txt'):
        df = pd.read_csv(input_path, sep='\t', low_memory=False)
    elif input_path.endswith('.parquet'):
        df = pd.read_parquet(input_path)
    else:
        raise ValueError("Định dạng file không được hỗ trợ. Vui lòng dùng .csv, .tsv, hoặc .parquet")
    
    initial_rows = len(df)
    print(f"[1] Số lượng ban đầu: {initial_rows:,} dòng")

    # 2. CHUẨN HÓA TÊN CỘT (Không phân biệt hoa thường)
    col_mapping = {}
    for col in df.columns:
        col_lower = col.strip().lower()
        if col_lower == 'chrom' or col_lower == 'chromosome':
            col_mapping[col] = 'CHROM'
        elif col_lower == 'pos' or col_lower == 'position':
            col_mapping[col] = 'POS'
        elif col_lower == 'ref' or col_lower == 'reference':
            col_mapping[col] = 'REF'
        elif col_lower == 'alt' or col_lower == 'alternate':
            col_mapping[col] = 'ALT'
        elif col_lower == 'label' or col_lower == 'clinicalsignificance' or col_lower == 'class':
            col_mapping[col] = 'Label'
            
    df = df.rename(columns=col_mapping)
    
    # Kiểm tra xem có đủ 5 cột bắt buộc chưa
    required_cols = ['CHROM', 'POS', 'REF', 'ALT', 'Label']
    missing_cols = [c for c in required_cols if c not in df.columns]
    if missing_cols:
        raise ValueError(f"Dữ liệu gốc thiếu các cột: {missing_cols}")

    # Xóa các dòng bị NaN ở các cột tọa độ cốt lõi
    df = df.dropna(subset=['CHROM', 'POS', 'REF', 'ALT', 'Label'])
    
    # 3. CHUẨN HÓA CHROMOSOME (Lọc 1-24, X, Y)
    # Loại bỏ tiền tố 'chr' nếu có, chuyển thành in hoa
    df['CHROM'] = df['CHROM'].astype(str).str.replace(r'^chr', '', regex=True, case=False).str.strip().str.upper()
    
    # Danh sách Chrom hợp lệ
    valid_chroms = [str(i) for i in range(1, 25)] + ['X', 'Y']
    df = df[df['CHROM'].isin(valid_chroms)]
    
    # 4. CHUẨN HÓA CỘT LABEL (Chuyển về 0 và 1)
    # Định nghĩa các từ khóa (viết thường)
    patho_terms = ['1', '1.0', 'pathogenic', 'likely pathogenic', 'pathogenic/likely pathogenic', 'patho', 'p', 'lp']
    benign_terms = ['0', '0.0', 'benign', 'likely benign', 'benign/likely benign', 'b', 'lb']
    
    # Ép kiểu về chuỗi, viết thường để so khớp
    label_str = df['Label'].astype(str).str.strip().str.lower()
    
    # Tạo mask (Mặt nạ)
    is_patho = label_str.isin(patho_terms)
    is_benign = label_str.isin(benign_terms)
    
    # Lọc bỏ những dòng có Label kỳ lạ (như VUS, Uncertain significance...)
    df = df[is_patho | is_benign].copy()
    
    # Cập nhật lại cột Label chính thức: Patho = 1, Benign = 0
    df['Label'] = np.where(df['Label'].astype(str).str.strip().str.lower().isin(patho_terms), 1, 0)
    
    # 5. CHUẨN HÓA POS VÀ TẠO VARIANT_ID
    # Xử lý trường hợp POS bị dính số thập phân (ví dụ 1014401.0)
    df['POS'] = df['POS'].astype(float).astype('Int64').astype(str)
    df['REF'] = df['REF'].astype(str).str.upper().str.strip()
    df['ALT'] = df['ALT'].astype(str).str.upper().str.strip()
    
    # Tạo Variant_ID theo chuẩn CHROM_POS_REF_ALT
    df['Variant_ID'] = df['CHROM'] + '_' + df['POS'] + '_' + df['REF'] + '_' + df['ALT']
    
    # Đưa các cột cốt lõi lên đầu DataFrame cho đẹp
    cols_order = ['Variant_ID', 'CHROM', 'POS', 'REF', 'ALT', 'Label']
    other_cols = [c for c in df.columns if c not in cols_order]
    df = df[cols_order + other_cols]
    
    # 6. THỐNG KÊ VÀ LƯU FILE
    final_rows = len(df)
    dropped_rows = initial_rows - final_rows
    patho_count = df['Label'].sum()
    benign_count = final_rows - patho_count
    
    print(f"[2] Đã xóa: {dropped_rows:,} dòng (thiếu data, sai Chromosome, hoặc Label không rõ ràng)")
    print(f"[3] Số lượng hợp lệ cuối cùng: {final_rows:,} dòng")
    print(f"    - Pathogenic (1): {patho_count:,} ({(patho_count/final_rows*100):.2f}%)")
    print(f"    - Benign (0): {benign_count:,} ({(benign_count/final_rows*100):.2f}%)")
    print(f"[4] Lấy mẫu Variant_ID: {df['Variant_ID'].iloc[0]}, {df['Variant_ID'].iloc[-1]}")
    
    print(f"[5] Đang xuất file Parquet...")
    df.to_parquet(output_parquet, index=False)
    print(f"✅ HOÀN TẤT! File đã lưu tại: {output_parquet}\n")
    
    return df
```

```python
import pandas as pd
import os
import warnings
warnings.filterwarnings('ignore')

def check_cross_dataset_leakage(df_base, df_target, base_name="Train", target_name="Benchmark"):
    """Hàm kiểm tra overlap như cũ của bạn"""
    print(f"\n🔍 ĐANG QUÉT RÒ RỈ: [{base_name}] vs [{target_name}]")
    
    base_vars, target_vars = set(df_base['Variant_ID'].dropna()), set(df_target['Variant_ID'].dropna())
    
    overlap_vars = base_vars.intersection(target_vars)
    
    print(f"   - Variant trùng lặp : {len(overlap_vars):,} biến thể")
    
    return {'overlap_variants': overlap_vars}


def clean_and_prepare_benchmark_for_vep(df_train, benchmark_datasets, output_dir="D:/variant_data/Benchmark_Ready"):
    """
    Hàm xử lý hàng loạt các tập Benchmark:
    1. Xóa overlap với Train.
    2. Chỉ giữ CHROM 10, 15, 20, Y.
    3. Sắp xếp lại.
    4. Báo cáo & Xuất file.
    """
    # Tạo thư mục chứa file đầu ra nếu chưa có
    os.makedirs(output_dir, exist_ok=True)
    
    # Chromosome đích muốn giữ lại cho tập Test/Benchmark
    target_chroms = ['10', '15', '20', 'Y']
    
    # Mapping để sắp xếp Chromosome
    chrom_order = {str(i): i for i in range(1, 23)}
    chrom_order.update({'X': 23, 'Y': 24, 'MT': 25, 'M': 25})

    for bench_name, df_bench in benchmark_datasets.items():
        print(f"\n{'='*70}")
        print(f"🚀 BẮT ĐẦU XỬ LÝ TẬP BENCHMARK: {bench_name}")
        print(f"{'='*70}")
        
        initial_len = len(df_bench)
        
        # ==========================================
        # BƯỚC 1: TÌM VÀ XÓA OVERLAP VỚI TẬP TRAIN
        # ==========================================
        overlaps = check_cross_dataset_leakage(df_train, df_bench, base_name="Train_Bo1", target_name=bench_name)
        
        # Cắt bỏ Variant trùng
        if len(overlaps['overlap_variants']) > 0:
            df_bench = df_bench[~df_bench['Variant_ID'].isin(overlaps['overlap_variants'])]
            
        print(f"   => Đã xóa bỏ các dữ liệu rò rỉ. Số lượng còn lại: {len(df_bench):,} / {initial_len:,}")

        # ==========================================
        # BƯỚC 2: CHỈ GIỮ LẠI CHROMOSOME ĐÍCH
        # ==========================================
        df_bench['CHROM'] = df_bench['CHROM'].astype(str).str.replace(r'^chr', '', regex=True, case=False).str.upper()
        df_bench = df_bench[df_bench['CHROM'].isin(target_chroms)].copy()
        print(f"   => Đã lọc giữ lại CHROM {target_chroms}. Số lượng còn lại: {len(df_bench):,}")
        
        # ==========================================
        # BƯỚC 3: SẮP XẾP CHUẨN SINH HỌC
        # ==========================================
        # Ép POS về dạng số để sắp xếp không bị lỗi (VD: 2 không bị xếp sau 10)
        df_bench['POS_num'] = pd.to_numeric(df_bench['POS'], errors='coerce')
        df_bench['CHROM_num'] = df_bench['CHROM'].map(chrom_order).fillna(99)
        
        # Sort theo NST -> Position
        df_bench = df_bench.sort_values(by=['CHROM_num', 'POS_num'], ascending=[True, True])
        
        # Dọn dẹp cột tạm
        df_bench = df_bench.drop(columns=['CHROM_num', 'POS_num']).reset_index(drop=True)

        # ==========================================
        # BƯỚC 4: BÁO CÁO TỔNG QUAN CHUẨN BỊ CHO VEP
        # ==========================================
        total_variants = len(df_bench)
        if total_variants == 0:
            print(f"⚠️ CẢNH BÁO: Tập {bench_name} TRỐNG sau khi lọc!")
            continue
            
        patho_count = df_bench['Label'].sum() if 'Label' in df_bench.columns else 0
        benign_count = total_variants - patho_count
        
        print(f"\n📊 BÁO CÁO DỮ LIỆU ĐÃ LÀM SẠCH (READY FOR VEP):")
        print(f"   - Tổng số Variant : {total_variants:,}")
        print(f"   - Chromosomes     : {df_bench['CHROM'].unique().tolist()}")
        if 'Label' in df_bench.columns:
            print(f"   - Phân phối Nhãn  : Pathogenic = {patho_count:,} ({(patho_count/total_variants*100):.2f}%) | Benign = {benign_count:,} ({(benign_count/total_variants*100):.2f}%)")

        # ==========================================
        # BƯỚC 5: XUẤT FILE PARQUET
        # ==========================================
        output_file = os.path.join(output_dir, f"{bench_name}_Cleaned_VEP.parquet")
        df_bench.to_parquet(output_file, index=False)
        print(f"✅ Đã xuất file thành công: {output_file}")
```
