```python
import pandas as pd
import numpy as np

# 1. ĐỌC DỮ LIỆU
# Thay bằng đường dẫn file Parquet đã có VEP của bạn
df = pd.read_parquet(r"C:\Users\dotru\STUDIE\FPTU\AiTA_Lab\data\ver3\clinvar_vep_mapping_final_with_date.parquet")

df
```

```python
# 2. TẠO CỘT LABEL TỪ CLINICAL SIGNIFICANCE
# Định nghĩa danh sách các nhãn
benign_terms = ['Benign/Likely benign', 'Benign', 'Likely benign']
patho_terms = ['Likely Pathogenic', 'Pathogenic', 'Pathogenic/Likely pathogenic']

# Lọc chỉ giữ lại những dòng thuộc 2 nhóm trên
df = df[df['ClinicalSignificance'].isin(benign_terms + patho_terms)].copy()

# Tạo cột Label (1 cho Patho, 0 cho Benign)
df['Label'] = np.where(df['ClinicalSignificance'].isin(patho_terms), 1, 0)

df
```

```python
# ==========================================
# 3. CHẠY EDA (EXPLORATORY DATA ANALYSIS)
# ==========================================

print("=== 1. TỔNG QUAN TOÀN BỘ DATASET ===")
total_variants = len(df)
patho_count = df['Label'].sum()
benign_count = total_variants - patho_count
print(f"Tổng số Variant: {total_variants:,}")
print(f"Pathogenic (1): {patho_count:,} ({patho_count/total_variants*100:.2f}%)")
print(f"Benign (0): {benign_count:,} ({benign_count/total_variants*100:.2f}%)")
print(f"Tổng số Gene (VEP SYMBOL) unique: {df['SYMBOL'].nunique():,}")
print("-" * 50)
```

```python
print("\n=== 2. THỐNG KÊ THEO REVIEW STATUS (CHẤT LƯỢNG NHÃN) ===")
review_stats = df.groupby('ReviewStatus').agg(
    Total_Variants=('Variant_ID', 'count'),
    Pathogenic_1=('Label', 'sum')
).reset_index()

review_stats['Benign_0'] = review_stats['Total_Variants'] - review_stats['Pathogenic_1']
review_stats['Patho_Ratio (%)'] = (review_stats['Pathogenic_1'] / review_stats['Total_Variants'] * 100).round(2)
# Sắp xếp theo lượng variant giảm dần
review_stats = review_stats.sort_values('Total_Variants', ascending=False)
print(review_stats.to_string(index=False))
print("-" * 50)
```

```python
print("\n=== 3. THỐNG KÊ THEO CHROMOSOME ===")
# Thống kê số lượng theo từng Chromosome
chr_stats = df.groupby('CHROM').agg(
    Total_Variants=('Variant_ID', 'count'),
    Pathogenic_1=('Label', 'sum')
).reset_index()

chr_stats['Benign_0'] = chr_stats['Total_Variants'] - chr_stats['Pathogenic_1']
chr_stats['Patho_Ratio (%)'] = (chr_stats['Pathogenic_1'] / chr_stats['Total_Variants'] * 100).round(2)
# Sắp xếp theo lượng variant giảm dần
chr_stats = chr_stats.sort_values('Total_Variants', ascending=False)
print(chr_stats.to_string(index=False))
print("-" * 50)
```

```python
print("\n=== 4. THỐNG KÊ THEO GENE ===")
gene_stats = df.groupby('SYMBOL').agg(
    Total_Variants=('Variant_ID', 'count'),
    Pathogenic_1=('Label', 'sum')
).reset_index()

gene_stats['Benign_0'] = gene_stats['Total_Variants'] - gene_stats['Pathogenic_1']
# Lọc ra các Gene có cả 2 nhãn (Phục vụ cho Bộ 2 & 3 sau này)
gene_stats['Has_Both_Labels'] = (gene_stats['Pathogenic_1'] > 0) & (gene_stats['Benign_0'] > 0)

gene_stats_sorted = gene_stats.sort_values('Total_Variants', ascending=False)
print(gene_stats_sorted.to_string(index=False))
```

```python
def stats_last_evaluated_year(df):
    print("============================================================")
    print("📅 THỐNG KÊ SỐ LƯỢNG BIẾN THỂ THEO NĂM ĐÁNH GIÁ (LastEvaluated)")
    print("============================================================")
    
    # 1. Chuyển đổi an toàn sang kiểu Datetime và trích xuất Năm (Year)
    # errors='coerce' sẽ tự động biến các chuỗi lỗi hoặc không hợp lệ thành giá trị rỗng (NaT)
    year_series = pd.to_datetime(df['LastEvaluated'], errors='coerce').dt.year
    
    # 2. Đếm số lượng, loại bỏ NaN và ép kiểu về số nguyên (Int64 để không bị đuôi .0)
    year_stats = year_series.value_counts().reset_index()
    year_stats.columns = ['Năm', 'Số lượng Variant']
    year_stats['Năm'] = year_stats['Năm'].astype('Int64')
    
    # 3. Sắp xếp theo Năm giảm dần (từ năm mới nhất về cũ nhất)
    year_stats = year_stats.sort_values('Năm', ascending=False)
    
    # 4. Tính toán tỷ lệ phần trăm
    total_variants = len(df)
    total_has_date = year_stats['Số lượng Variant'].sum()
    missing_date = total_variants - total_has_date
    
    year_stats['Tỷ lệ (trong nhóm có ngày)'] = (year_stats['Số lượng Variant'] / total_has_date * 100).round(2).astype(str) + '%'
    
    # 5. In báo cáo tổng quan
    print(f"- Tổng số biến thể trong Dataset: {total_variants:,}")
    print(f"- Số biến thể CÓ ngày đánh giá hợp lệ: {total_has_date:,} ({(total_has_date/total_variants*100):.2f}%)")
    print(f"- Số biến thể TRỐNG ngày (NaN hoặc lỗi): {missing_date:,} ({(missing_date/total_variants*100):.2f}%)")
    print("-" * 60)
    
    # In bảng chi tiết theo năm
    print(year_stats.to_string(index=False))
    
    return year_stats

# =========================================================
# GỌI HÀM KIỂM TRA TRÊN DATAFRAME TỔNG CỦA BẠN
# =========================================================
df_year_stats = stats_last_evaluated_year(df)
```

```python
# 2. MÃ HÓA ĐỘ TIN CẬY REVIEW STATUS (DÙNG CHO BỘ 3)
print("[2/5] Đang đánh giá Rank cho ReviewStatus...")
review_priority = {
    'practice guideline': 5,
    'reviewed by expert panel': 4,
    'criteria provided, multiple submitters, no conflicts': 3,
    'criteria provided, single submitter': 2,
    'criteria provided, conflicting classifications': 1
}
# Các loại 'no classification', 'no assertion' sẽ mặc định nhận rank 0
df['Review_Rank'] = df['ReviewStatus'].map(review_priority).fillna(0)

# =========================================================
# THUẬT TOÁN GREEDY CHIA CHROMOSOME (ĐẢM BẢO KHÔNG TRÙNG LẶP)
# =========================================================
print("[3/5] Đang chạy thuật toán phân bổ Chromosome (80-10-10)...")
chr_stats = df['CHROM'].value_counts().reset_index()
chr_stats.columns = ['CHROM', 'Count']

total_variants = chr_stats['Count'].sum()
target_train = total_variants * 0.80
target_val = total_variants * 0.10
target_test = total_variants * 0.10

splits = {'Train': [], 'Val': [], 'Test': []}
current_counts = {'Train': 0, 'Val': 0, 'Test': 0}
targets = {'Train': target_train, 'Val': target_val, 'Test': target_test}

# Lặp qua các Chromosome từ lớn nhất đến nhỏ nhất
for _, row in chr_stats.iterrows():
    chrom = row['CHROM']
    count = row['Count']
    
    # Tìm rương đang thiếu hụt nhiều nhất so với target
    deficits = {k: targets[k] - current_counts[k] for k in targets}
    best_split = max(deficits, key=deficits.get)
    
    # Gán Chromosome vào rương đó
    splits[best_split].append(chrom)
    current_counts[best_split] += count

print("\n--- KẾT QUẢ CHIA CHROMOSOME ---")
for k in splits:
    pct = current_counts[k] / total_variants * 100
    print(f"{k} ({pct:.2f}%): {splits[k]}")
print("--------------------------------\n")

# Gán cột Split vào DataFrame tổng
def get_split(chrom):
    if chrom in splits['Train']: return 'Train'
    elif chrom in splits['Val']: return 'Val'
    else: return 'Test'

df['Split'] = df['CHROM'].apply(get_split)

# =========================================================
# TẠO BỘ 1: CHỈ CHIA TRAIN/VAL/TEST THEO CHROMOSOME
# =========================================================
print("[4/5] Đang xuất Bộ 1...")
df_bo1 = df.copy()

# =========================================================
# TẠO BỘ 2: BỘ 1 + LỌC BỎ GENE CHỈ CÓ 1 NHÃN
# =========================================================
print("Đang xuất Bộ 2...")
# Đếm số lượng nhãn độc nhất của mỗi Gene
gene_label_counts = df_bo1.groupby('SYMBOL')['Label'].nunique()
# Giữ lại các Gene có đúng 2 nhãn (cả 0 và 1)
valid_genes = gene_label_counts[gene_label_counts == 2].index
df_bo2 = df_bo1[df_bo1['SYMBOL'].isin(valid_genes)].copy()

# =========================================================
# TẠO BỘ 3: BỘ 2 + DOWNSAMPLING 50:50 DỰA THEO REVIEW RANK
# =========================================================
print("Đang xuất Bộ 3 (Downsampling 50:50 thông minh)...")

# Bước A: Sắp xếp ưu tiên: Gene -> Label -> Rank ReviewStatus (Giảm dần)
df_bo3_temp = df_bo2.sort_values(by=['SYMBOL', 'Label', 'Review_Rank'], ascending=[True, True, False])

# Bước B: Đánh số thứ tự từng Variant trong cùng một nhóm (Gene + Label)
df_bo3_temp['Row_Num'] = df_bo3_temp.groupby(['SYMBOL', 'Label']).cumcount()

# Bước C: Đếm số lượng Patho và Benign của từng Gene
class_counts = df_bo3_temp.groupby(['SYMBOL', 'Label']).size().unstack(fill_value=0)
# Số lượng giới hạn = min(số lượng Patho, số lượng Benign)
class_counts['Min_Count'] = class_counts.min(axis=1)

# Bước D: Lọc bằng cách ghép (Merge) Min_Count vào data và cắt bỏ biến thể dư thừa
df_bo3 = df_bo3_temp.merge(class_counts[['Min_Count']], on='SYMBOL', how='left')
# Chỉ giữ lại các Variant có số thứ tự < Min_Count
df_bo3 = df_bo3[df_bo3['Row_Num'] < df_bo3['Min_Count']].copy()

# Dọn dẹp cột phụ
df_bo3 = df_bo3.drop(columns=['Row_Num', 'Min_Count'])
```

```python
def eda_dataset(df, name):
    print(f"============================================================")
    print(f"🌟 BÁO CÁO EDA CHI TIẾT: {name}")
    print(f"============================================================")
    
    # 1. TỔNG QUAN
    total_variants = len(df)
    total_genes = df['SYMBOL'].nunique()
    total_chroms = df['CHROM'].nunique()
    patho_count = df['Label'].sum()
    benign_count = total_variants - patho_count
    
    print("1. TỔNG QUAN KÍCH THƯỚC:")
    print(f"   - Tổng số biến thể: {total_variants:,}")
    print(f"   - Tổng số Gene: {total_genes:,}")
    print(f"   - Tổng số Chromosome: {total_chroms:,}")
    print(f"   - Phân phối nhãn: Pathogenic = {patho_count:,} ({patho_count/total_variants*100:.2f}%) | Benign = {benign_count:,} ({benign_count/total_variants*100:.2f}%)")
    
    # 2. KIỂM TRA ĐIỀU KIỆN LỌC GENE
    gene_label_counts = df.groupby('SYMBOL')['Label'].nunique()
    single_label_genes = (gene_label_counts == 1).sum()
    print("\n2. KIỂM TRA TÍNH HỢP LỆ CỦA GENE:")
    print(f"   - Số Gene chỉ có 1 nhãn (Patho hoặc Benign): {single_label_genes:,}")
    if "BỘ 1" not in name:
        print(f"     => Tình trạng: {'✅ PASS (Đạt 0 như yêu cầu)' if single_label_genes == 0 else '❌ FAIL'}")

    # 3. PHÂN BỔ THEO TẬP TRAIN / VAL / TEST
    print("\n3. PHÂN PHỐI DỮ LIỆU CHIA TẬP:")
    split_stats = df.groupby('Split').agg(
        Variants=('Variant_ID', 'count'),
        Genes=('SYMBOL', 'nunique'),
        Chromosomes=('CHROM', 'nunique'),
        Pathogenic=('Label', 'sum')
    ).reset_index()
    
    split_stats['Benign'] = split_stats['Variants'] - split_stats['Pathogenic']
    split_stats['% Tổng'] = (split_stats['Variants'] / total_variants * 100).round(2)
    split_stats['% Patho'] = (split_stats['Pathogenic'] / split_stats['Variants'] * 100).round(2)
    
    print(split_stats[['Split', 'Variants', '% Tổng', 'Pathogenic', 'Benign', '% Patho', 'Genes', 'Chromosomes']].to_string(index=False))

    # 4. KIỂM TRA CHỐNG RÒ RỈ DATA (DATA LEAKAGE)
    print("\n4. KIỂM TRA CHỐNG RÒ RỈ (LEAKAGE) TẬP DỮ LIỆU:")
    train_genes = set(df[df['Split'] == 'Train']['SYMBOL'])
    val_genes = set(df[df['Split'] == 'Val']['SYMBOL'])
    test_genes = set(df[df['Split'] == 'Test']['SYMBOL'])
    
    overlap_tv = len(train_genes.intersection(val_genes))
    overlap_tt = len(train_genes.intersection(test_genes))
    overlap_vt = len(val_genes.intersection(test_genes))
    
    total_overlaps = overlap_tv + overlap_tt + overlap_vt
    print(f"   - Trùng lặp Gene: Train-Val ({overlap_tv}) | Train-Test ({overlap_tt}) | Val-Test ({overlap_vt})")
    print(f"     => Tình trạng Leakage: {'✅ AN TOÀN (0 Trùng lặp)' if total_overlaps == 0 else '❌ BỊ LEAK DATA'}")

    # =======================================================
    # TÍNH NĂNG 5 MỚI: THỐNG KÊ NHÃN MỖI CHROMOSOME
    # =======================================================
    print("\n5. THỐNG KÊ NHÃN THEO TỪNG CHROMOSOME:")
    chr_stats = df.groupby('CHROM').agg(
        Total=('Variant_ID', 'count'),
        Pathogenic=('Label', 'sum')
    ).reset_index()
    chr_stats['Benign'] = chr_stats['Total'] - chr_stats['Pathogenic']
    chr_stats['% Patho'] = (chr_stats['Pathogenic'] / chr_stats['Total'] * 100).round(2)
    chr_stats = chr_stats.sort_values('Total', ascending=False)
    
    # In toàn bộ danh sách Chromosome
    print(chr_stats.to_string(index=False))

    # =======================================================
    # TÍNH NĂNG 6 MỚI: THỐNG KÊ NHÃN MỖI GENE
    # =======================================================
    print(f"\n6. THỐNG KÊ NHÃN THEO TỪNG GENE (HIỂN THỊ TOP 15 / {total_genes:,} GENE):")
    gene_stats = df.groupby('SYMBOL').agg(
        Total=('Variant_ID', 'count'),
        Pathogenic=('Label', 'sum')
    ).reset_index()
    gene_stats['Benign'] = gene_stats['Total'] - gene_stats['Pathogenic']
    gene_stats['% Patho'] = (gene_stats['Pathogenic'] / gene_stats['Total'] * 100).round(2)
    gene_stats = gene_stats.sort_values('Total', ascending=False)
    
    print(gene_stats.head(15).to_string(index=False))
    print("...\n\n")

# =========================================================
# GỌI HÀM KIỂM TRA
# =========================================================

# Chạy thử nghiệm trên 3 bộ dữ liệu
eda_dataset(df_bo1, "BỘ 1 (Chỉ chia Chromosome, giữ nguyên gốc)")
eda_dataset(df_bo2, "BỘ 2 (Lọc bỏ các Gene chỉ chứa 1 loại Nhãn)")
eda_dataset(df_bo3, "BỘ 3 (Downsampling ép tỷ lệ 50:50)")
```

```python
# 2. MÃ HÓA ĐỘ TIN CẬY REVIEW STATUS
print("[2/6] Đang đánh giá Rank cho ReviewStatus...")
review_priority = {
    'practice guideline': 5,
    'reviewed by expert panel': 4,
    'criteria provided, multiple submitters, no conflicts': 3,
    'criteria provided, single submitter': 2,
    'criteria provided, conflicting classifications': 1
}
df['Review_Rank'] = df['ReviewStatus'].map(review_priority).fillna(0)

# =========================================================
# XỬ LÝ THỜI GIAN (YEAR) TRƯỚC KHI CHIA CHROMOSOME
# =========================================================
print("[3/6] Đang trích xuất thời gian và loại bỏ biến thể không có ngày tháng...")
df['Year'] = pd.to_datetime(df['LastEvaluated'], errors='coerce').dt.year

# Bắt buộc drop các dòng không có năm để đảm bảo tính nghiêm ngặt của Time-split
df_time = df.dropna(subset=['Year']).copy()

# =========================================================
# THUẬT TOÁN GREEDY THÔNG MINH (TIME-AWARE CHROMOSOME SPLIT)
# =========================================================
print("[4/6] Đang chạy thuật toán phân bổ Chromosome (80-10-10 kết hợp Thời gian)...")

# Tính "sản lượng" biến thể hợp lệ của mỗi Chromosome nếu đưa vào từng Rương
chr_stats = []
for chrom, group in df_time.groupby('CHROM'):
    chr_stats.append({
        'CHROM': chrom,
        'Yield_Train': (group['Year'] <= 2024).sum(),
        'Yield_Val': (group['Year'] == 2025).sum(),
        'Yield_Test': (group['Year'] >= 2026).sum(),
        'Total': len(group)
    })
chr_df = pd.DataFrame(chr_stats).sort_values('Total', ascending=False)

# Ước lượng tổng số variant sẽ giữ lại được (để tính target 80-10-10)
estimated_kept = chr_df[['Yield_Train', 'Yield_Val', 'Yield_Test']].max(axis=1).sum()
targets = {
    'Train': estimated_kept * 0.80,
    'Val': estimated_kept * 0.10,
    'Test': estimated_kept * 0.10
}

splits = {'Train': [], 'Val': [], 'Test': []}
current_counts = {'Train': 0, 'Val': 0, 'Test': 0}

for _, row in chr_df.iterrows():
    chrom = row['CHROM']
    yields = {
        'Train': row['Yield_Train'], 
        'Val': row['Yield_Val'], 
        'Test': row['Yield_Test']
    }
    
    # Tính mức độ thiếu hụt của các rương
    deficits = {k: targets[k] - current_counts[k] for k in targets}
    
    # Sắp xếp rương theo độ "đói" dữ liệu giảm dần
    sorted_splits = sorted(deficits.keys(), key=lambda k: deficits[k], reverse=True)
    
    # Chọn rương "đói" nhất MÀ CÓ THỂ CUNG CẤP DỮ LIỆU (> 0)
    best_split = sorted_splits[0]
    for s in sorted_splits:
        if yields[s] > 0:
            best_split = s
            break
            
    # Gán Chromosome vào rương đã chọn
    splits[best_split].append(chrom)
    current_counts[best_split] += yields[best_split]

# =========================================================
# LỌC NGHIÊM NGẶT THEO ĐIỀU KIỆN THỜI GIAN
# =========================================================
def get_split(chrom):
    if chrom in splits['Train']: return 'Train'
    elif chrom in splits['Val']: return 'Val'
    elif chrom in splits['Test']: return 'Test'
    else: return 'Drop'

df_time['Temp_Split'] = df_time['CHROM'].apply(get_split)

# CHỈ GIỮ LẠI CÁC BIẾN THỂ THỎA MÃN CHÍNH XÁC NĂM CỦA RƯƠNG ĐÓ (Bước này xén bớt data thừa)
cond_train = (df_time['Temp_Split'] == 'Train') & (df_time['Year'] <= 2024)
cond_val   = (df_time['Temp_Split'] == 'Val')   & (df_time['Year'] == 2025)
cond_test  = (df_time['Temp_Split'] == 'Test')  & (df_time['Year'] >= 2026)

# Đây là DataFrame KHỞI NGUỒN CHÍNH THỨC, đã sạch sẽ 100% về không gian và thời gian
df_filtered = df_time[cond_train | cond_val | cond_test].copy()
df_filtered.rename(columns={'Temp_Split': 'Split'}, inplace=True)

print("\n--- KẾT QUẢ CHIA CHROMOSOME KẾT HỢP TIME-SPLIT ---")
final_total = len(df_filtered)
for k in splits:
    actual_count = len(df_filtered[df_filtered['Split'] == k])
    pct = actual_count / final_total * 100 if final_total > 0 else 0
    print(f"{k} ({pct:.2f}% | {actual_count:,} variants): Chromosomes {splits[k]}")
print(f"Tổng số Variant hợp lệ giữ lại: {final_total:,} (Mất {len(df) - final_total:,} do thiếu ngày hoặc khác năm)")
print("--------------------------------------------------\n")

# =========================================================
# TẠO BỘ 1: GỐC (Đã xén theo Time & Chromosome)
# =========================================================
print("[5/6] Đang xuất các Bộ dữ liệu...")
df_bo1 = df_filtered.copy()

# =========================================================
# TẠO BỘ 2: LỌC BỎ GENE CHỈ CÓ 1 NHÃN
# =========================================================
gene_label_counts = df_bo1.groupby('SYMBOL')['Label'].nunique()
valid_genes = gene_label_counts[gene_label_counts == 2].index
df_bo2 = df_bo1[df_bo1['SYMBOL'].isin(valid_genes)].copy()

# =========================================================
# TẠO BỘ 3: DOWNSAMPLING 50:50 THEO REVIEW RANK
# =========================================================
# Bước A: Sắp xếp
df_bo3_temp = df_bo2.sort_values(by=['SYMBOL', 'Label', 'Review_Rank'], ascending=[True, True, False])

# Bước B: Đánh số thứ tự
df_bo3_temp['Row_Num'] = df_bo3_temp.groupby(['SYMBOL', 'Label']).cumcount()

# Bước C: Tính giới hạn Min_Count
class_counts = df_bo3_temp.groupby(['SYMBOL', 'Label']).size().unstack(fill_value=0)
class_counts['Min_Count'] = class_counts.min(axis=1)

# Bước D: Ghép và Lọc
df_bo3 = df_bo3_temp.merge(class_counts[['Min_Count']], on='SYMBOL', how='left')
df_bo3 = df_bo3[df_bo3['Row_Num'] < df_bo3['Min_Count']].copy()

# Dọn dẹp
df_bo3 = df_bo3.drop(columns=['Row_Num', 'Min_Count'])
```

```python
def check_cross_dataset_leakage(df_base, df_target, base_name="Train/Val", target_name="Benchmark Test"):
    """
    Hàm so khớp và kiểm tra rò rỉ dữ liệu giữa 2 tập dataset trên 3 cấp độ:
    Variant_ID, SYMBOL (Gene), CHROM.
    
    Trả về dictionary chứa các tập hợp trùng lặp để bạn có thể dùng để filter (nếu cần).
    """
    print(f"============================================================")
    print(f"🔍 KIỂM TRA RÒ RỈ: [{base_name}] vs [{target_name}]")
    print(f"============================================================")
    
    # Kiểm tra xem các cột cần thiết có tồn tại không
    required_cols = ['Variant_ID', 'SYMBOL', 'CHROM']
    for col in required_cols:
        if col not in df_base.columns or col not in df_target.columns:
            print(f"[LỖI] Cả 2 tập dữ liệu cần phải có cột '{col}'")
            return None

    # 1. Trích xuất các tập hợp (Sets)
    base_vars = set(df_base['Variant_ID'].dropna())
    target_vars = set(df_target['Variant_ID'].dropna())
    
    base_genes = set(df_base['SYMBOL'].dropna())
    target_genes = set(df_target['SYMBOL'].dropna())
    
    base_chroms = set(df_base['CHROM'].astype(str).dropna())
    target_chroms = set(df_target['CHROM'].astype(str).dropna())
    
    # 2. Tìm điểm giao (Intersection)
    overlap_vars = base_vars.intersection(target_vars)
    overlap_genes = base_genes.intersection(target_genes)
    overlap_chroms = base_chroms.intersection(target_chroms)
    
    # 3. In báo cáo
    print(f"1. Cấp độ Variant_ID:")
    print(f"   - Tập {base_name}: {len(base_vars):,} variants")
    print(f"   - Tập {target_name}: {len(target_vars):,} variants")
    if len(overlap_vars) == 0:
        print(f"   => Tình trạng: ✅ AN TOÀN (0 trùng lặp)")
    else:
        print(f"   => Tình trạng: ❌ LEAKAGE ({len(overlap_vars):,} variants trùng lặp)")
        # In thử 3 ID trùng lặp để kiểm tra
        print(f"      (Ví dụ: {list(overlap_vars)[:3]})")

    print(f"\n2. Cấp độ SYMBOL (Gene):")
    print(f"   - Tập {base_name}: {len(base_genes):,} genes")
    print(f"   - Tập {target_name}: {len(target_genes):,} genes")
    if len(overlap_genes) == 0:
        print(f"   => Tình trạng: ✅ AN TOÀN (0 trùng lặp)")
    else:
        print(f"   => Tình trạng: ❌ LEAKAGE ({len(overlap_genes):,} genes trùng lặp)")
        print(f"      (Ví dụ: {list(overlap_genes)[:5]})")

    print(f"\n3. Cấp độ CHROM (Nhiễm sắc thể):")
    print(f"   - Tập {base_name}: {len(base_chroms)} chroms")
    print(f"   - Tập {target_name}: {len(target_chroms)} chroms")
    if len(overlap_chroms) == 0:
        print(f"   => Tình trạng: ✅ AN TOÀN (0 trùng lặp)")
    else:
        print(f"   => Tình trạng: ❌ LEAKAGE ({len(overlap_chroms)} chroms trùng lặp)")
        print(f"      (Danh sách: {overlap_chroms})")
        
    print("\n")
    
    # Trả về các set trùng lặp để bạn có thể code loại bỏ chúng nếu muốn
    return {
        'overlap_variants': overlap_vars,
        'overlap_genes': overlap_genes,
        'overlap_chroms': overlap_chroms
    }

# =========================================================
# VÍ DỤ CÁCH SỬ DỤNG CHO NHIỀU TẬP BENCHMARK
# =========================================================

# Giả sử bạn có 1 dict chứa các tập dataset đọc từ parquet
# benchmark_datasets = {
#     "Benchmark_1_ClinVar_2024": pd.read_parquet("bench1.parquet"),
#     "Benchmark_2_VarSome": pd.read_parquet("bench2.parquet"),
#     "Benchmark_3_Literature": pd.read_parquet("bench3.parquet")
# }

# Giả sử df_bo3_train_val là tập bạn đã gộp từ Train và Val của Bộ 3
# df_bo3_train_val = pd.concat([df_bo3[df_bo3['Split'] == 'Train'], df_bo3[df_bo3['Split'] == 'Val']])

# Chạy vòng lặp kiểm tra toàn bộ
# for bench_name, df_bench in benchmark_datasets.items():
#     overlaps = check_cross_dataset_leakage(
#         df_base=df_bo3_train_val, 
#         df_target=df_bench, 
#         base_name="Train+Val (Bộ 3)", 
#         target_name=bench_name
#     )
    
#     # Nếu phát hiện trùng variant, bạn có thể tự động TẨY RỬA tập Benchmark như sau:
#     # if len(overlaps['overlap_variants']) > 0:
#     #     df_bench_clean = df_bench[~df_bench['Variant_ID'].isin(overlaps['overlap_variants'])]
#     #     print(f"Đã dọn dẹp xong tập {bench_name}, còn lại {len(df_bench_clean)} variants sạch.")
```
